```python
import pandas as pd

# === 输入文件路径 ===
meer_path = "MeerKAT.csv"
fast_path = "FAST.csv"
output_path = "Combined_Pulsars.csv"

# === 读取两个CSV ===
meer_df = pd.read_csv(meer_path)
fast_df = pd.read_csv(fast_path)

# === 添加来源列（如果已有Telescope列，可以直接保留，不需要额外添加） ===
meer_df["Telescope"] = "MeerKAT"
fast_df["Telescope"] = "FAST"

# === 合并数据 ===
combined_df = pd.concat([meer_df, fast_df], ignore_index=True)

# === 保存合并文件 ===
combined_df.to_csv(output_path, index=False)
print(f"✅ 已生成合并文件: {output_path}")

# === 统计两个文件相同的 PSR ===
meer_psrs = set(meer_df["PSR"].unique())
fast_psrs = set(fast_df["PSR"].unique())

# 求交集
common_psrs = meer_psrs.intersection(fast_psrs)

# 输出统计
print("📊 两个文件中相同的 PSR 共:", len(common_psrs), "个")
print("示例前10个:", list(common_psrs)[:10])

# 如果想保存为CSV
common_df = pd.DataFrame({"PSR": sorted(common_psrs)})
common_df.to_csv("Common_PSRs.csv", index=False)
print("✅ 已生成相同PSR列表: Common_PSRs.csv")
```
