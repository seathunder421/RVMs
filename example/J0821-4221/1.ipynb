```python
import pickle
import numpy as np

# === 载入数据 ===
filename = "J0821-4221_0_RVM_fit_dynesty.pickle"
data = pickle.load(open(filename, "rb"))

# === 提取最大似然对应的 index ===
logl = np.array(data.logl)  # 对数似然数组
ml_idx = np.argmax(logl)    # 最大似然对应的 index

# === 提取参数 ===
# 你可能会看到 data.samples 或 data.results['samples']，根据结构稍作调整
try:
    samples = np.array(data.samples)  # dynesty sampler 中的参数点
except AttributeError:
    samples = np.array(data["samples"])  # 如果是字典结构（老版本）

# === 最大似然参数组 ===
ml_params = samples[ml_idx]
print("最大似然对应参数组为：", ml_params)
```
